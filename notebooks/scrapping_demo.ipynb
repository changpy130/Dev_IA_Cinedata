```python
import requests
from bs4 import BeautifulSoup
```

```python
URL_WIKI = "https://en.wikipedia.org/wiki/Spider-Man:_Brand_New_Day"
```

### Requests

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0 Safari/537.36"
}

response = requests.get(URL_WIKI, headers=headers)

print(response.status_code)
```

### Soup

```python
soup = BeautifulSoup(response.text, "html.parser")
print(soup.title)
```

#### Find h1

```python
h1 = soup.find("h1")
print(h1.text)
```

#### Find infobox

```python
infobox = soup.find("table", {"class": "infobox"})
print(infobox.text)
```

```python
rows = infobox.find_all("tr")
print(len(rows))
```

```python
movie_infobox = {}

for row in rows:
    header = row.find("th")
    data = row.find("td")

    if header and data:
        items = data.find_all("li")

        if items:
            value = ", ".join(li.text.strip() for li in items)
        else:
            value = data.text.strip()

        # print(f"{header.text.strip()} -> {value}")
        movie_infobox[header.text.strip()] = value

movie_infobox_clean = { 
    key: value.replace("\xa0", " ") for key, value in movie_infobox.items() 
}

print(movie_infobox_clean)
```

#### Find plot

```python
plot_head = soup.find(id="Plot")
print(plot_head)
```

```python
plot_div = plot_head.parent

for sibling in plot_div.find_next_siblings():
    print(sibling)
```

```python
paragraphs = plot_div.find_next_siblings("p")
plot_text = "\n\n".join(p.text for p in paragraphs)
print(plot_text)
```
