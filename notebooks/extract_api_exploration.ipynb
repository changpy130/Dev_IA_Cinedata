```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv
```

## Load Config

```python
load_dotenv()
```

```python
TMDB_API_KEY = os.getenv("TMDB_API_KEY")
TMDB_BASE_URL = os.getenv("TMDB_BASE_URL")
```

## APIs

### Popular Movies

```python
def get_popular_movies(page=1, language="en-UK"):
    url = f"{TMDB_BASE_URL}/movie/popular"

    params = {
        "api_key": TMDB_API_KEY,
        "page": page,
        "language": language
    }

    response = requests.get(url, params=params)
    return response.json()
```

### Genre Mapping

```python
def get_genre_mapping(language="en-UK"):
    url = f"{TMDB_BASE_URL}/genre/movie/list"

    params = {
        "api_key": TMDB_API_KEY,
        "language": language
    }

    response = requests.get(url, params=params)
    return response.json()
```

### Movie Details

```python
def get_movie_details(movie_id, language="en-UK"):
    url = f"{TMDB_BASE_URL}/movie/{movie_id}"

    params = {
        "api_key": TMDB_API_KEY,
        "language": language
    }

    response = requests.get(url, params=params)
    return response.json()
```

### Cast

```python
def get_movie_cast(movie_id, language="en-UK"):
    url = f"{TMDB_BASE_URL}/movie/{movie_id}/credits"

    params = {
        "api_key": TMDB_API_KEY,
        "language": language
    }

    response = requests.get(url, params=params)
    return response.json()
```

## Into DataFrame

```python
popular_list = get_popular_movies()
df_popular_movies = pd.DataFrame(popular_list['results'])
df_popular_movies
```

## Mapping the genre

```python
genre_list = get_genre_mapping()
genre_dict = { g['id']: g['name'] for g in genre_list['genres'] }
genre_dict
```

```python
df_popular_movies['genre_names'] = df_popular_movies.genre_ids.apply(
    lambda list: [genre_dict.get(i) for i in list]
)

df_popular_movies
```

## Movie details in DataFrame

```python
details_list = []

for id in df_popular_movies['id'].head(5):
    detail = get_movie_details(id)
    details_list.append(detail)

df_details = pd.DataFrame(details_list)
df_details
```

## Add Cast into main df

```python
cast_list = []

for id in df_popular_movies.id:
    cast = get_movie_cast(id)
    top_5 = cast['cast'][:5]
    cast_list.append({
        'cast': [{'name': i['name'], 'character': i['character']} for i in top_5]
    })

df_popular_movies['cast'] = cast_list
```
